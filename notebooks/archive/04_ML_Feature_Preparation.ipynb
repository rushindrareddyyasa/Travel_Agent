```python
## Import standard scaler
# StandardScaler converts numerical features to a common scale.
# This is important because our recommendation system
# will compare destinations based on feature similarity.

from sklearn.preprocessing import StandardScaler
```

```python
# Cosine similarity measures how similar two feature vectors are.
# We will use it to compare the user's travel preferences
# with every destination in our dataset.

from sklearn.metrics.pairwise import cosine_similarity
```

```python
def calculate_similarity(user_vector, destination_matrix):
    """
    Calculate how similar the user's preferences are
    to every destination in the dataset.

    Parameters:
        user_vector: User's feature vector
        destination_matrix: Scaled destination features

    Returns:
        Similarity score for every destination
    """

    # Calculate cosine similarity between the user's
    # preference vector and all destination feature vectors.
    similarity_scores = cosine_similarity(
        user_vector,
        destination_matrix
    )

    # Convert the result into a one-dimensional array.
    return similarity_scores[0]
```

```python
# Define a sample user's travel preferences.
# These values will later come from the actual user interface.

user_budget = 25000
user_duration = 4
user_season = "Nov-Feb"
user_category = "Adventure"
```

```python
# Calculate cosine similarity between the user's preferences
# and all 92 destination feature vectors.

user_similarity_scores = calculate_similarity(
    user_scaled,
    X_scaled.values
)

# Add the similarity scores to our recommendation dataset.
# We create a copy so the original dataset remains unchanged.

recommendation_results = recommendation_data.copy()

recommendation_results["Similarity_Score"] = user_similarity_scores

# Sort destinations from highest similarity to lowest similarity.
recommendation_results = recommendation_results.sort_values(
    by="Similarity_Score",
    ascending=False
)

# Display the top 10 recommended destinations.
recommendation_results[
    ["Destination", "State", "Category", "Best_Season", "Similarity_Score"]
].head(10)
```

```python
# Calculate how closely each destination's budget matches
# the user's requested budget.
#
# A score of 1 means the destination exactly matches
# the user's budget.
#
# As the difference increases, the score decreases.

budget_difference = (
    destination_df["Average_Budget"] - user_budget
).abs()

budget_score = 1 - (
    budget_difference / destination_df["Average_Budget"].max()
)

# Keep the score between 0 and 1.
budget_score = budget_score.clip(0, 1)

# Display the first few scores.
budget_score.head()
```

```python
# Calculate how closely each destination's recommended
# trip duration matches the user's requested duration.
#
# A score of 1 means the duration matches exactly.

duration_difference = (
    destination_df["Recommended_Trip_Duration"] - user_duration
).abs()

duration_score = 1 - (
    duration_difference /
    destination_df["Recommended_Trip_Duration"].max()
)

# Keep the score between 0 and 1.
duration_score = duration_score.clip(0, 1)

# Display the first few scores.
duration_score.head()
```

```python
# Give a binary score for the user's preferred season.
#
# 1 = destination's best season matches the user's season
# 0 = destination's best season does not match

season_score = (
    destination_df["Best_Season"] == user_season
).astype(int)

# Display the season matching scores.
season_score.head()
```

```python
# Give a binary score for the user's preferred category.
#
# 1 = destination belongs to the preferred category
# 0 = destination belongs to another category

category_score = (
    destination_df["Category"] == user_category
).astype(int)

# Display the category matching scores.
category_score.head()
```

```python
# Display all destinations that match the user's
# preferred category.

category_matches = destination_df[
    destination_df["Category"] == user_category
][
    ["Destination", "Category"]
]

# Show the matching destinations.
category_matches
```

```python
# Normalize the Adventure_Score to a 0-1 range.
#
# This allows us to combine it with our other scores,
# which are also between 0 and 1.

adventure_score = (
    destination_df["Adventure_Score"] -
    destination_df["Adventure_Score"].min()
) / (
    destination_df["Adventure_Score"].max() -
    destination_df["Adventure_Score"].min()
)

# Display the normalized adventure scores.
adventure_score.head()
```

```python
# Define the importance of each recommendation component.
# The weights add up to 1.0 (100%).

SEASON_WEIGHT = 0.30
CATEGORY_WEIGHT = 0.30
BUDGET_WEIGHT = 0.20
DURATION_WEIGHT = 0.10
ADVENTURE_WEIGHT = 0.10


# Calculate the final recommendation score.
#
# Each individual score is between 0 and 1.
# The weighted sum therefore also produces a score
# between 0 and 1.

recommendation_score = (
    season_score * SEASON_WEIGHT
    + category_score * CATEGORY_WEIGHT
    + budget_score * BUDGET_WEIGHT
    + duration_score * DURATION_WEIGHT
    + adventure_score * ADVENTURE_WEIGHT
)


# Create a copy of the destination information
# so we can attach the individual scores and final score.

scored_destinations = destination_df[
    [
        "Destination",
        "State",
        "Category",
        "Best_Season",
        "Average_Budget",
        "Recommended_Trip_Duration",
        "Adventure_Score"
    ]
].copy()


# Add each individual scoring component.
scored_destinations["Season_Score"] = season_score
scored_destinations["Category_Score"] = category_score
scored_destinations["Budget_Score"] = budget_score
scored_destinations["Duration_Score"] = duration_score
scored_destinations["Adventure_Score_Normalized"] = adventure_score


# Add the final weighted recommendation score.
scored_destinations["Recommendation_Score"] = recommendation_score


# Sort destinations from highest recommendation score
# to lowest recommendation score.

scored_destinations = scored_destinations.sort_values(
    by="Recommendation_Score",
    ascending=False
)


# Display the top 10 recommendations.
scored_destinations.head(10)
```

Recommendation function

```python
def recommend_destinations(
    budget,
    duration,
    season,
    category,
    top_n=5
):
    """
    Recommend travel destinations based on user preferences.

    Parameters:
        budget   : User's maximum/preferred travel budget.
        duration : Preferred number of travel days.
        season   : Preferred travel season.
        category : Preferred destination category.
        top_n    : Number of recommendations to return.

    Returns:
        A DataFrame containing the top recommended destinations.
    """

    # ---------------------------------------------------------
    # 1. Calculate budget similarity
    # ---------------------------------------------------------
    # A destination closer to the user's budget
    # receives a higher score.

    budget_difference = (
        destination_df["Average_Budget"] - budget
    ).abs()

    budget_score = (
        1 -
        budget_difference /
        destination_df["Average_Budget"].max()
    ).clip(0, 1)


    # ---------------------------------------------------------
    # 2. Calculate duration similarity
    # ---------------------------------------------------------
    # An exact duration match receives a score of 1.

    duration_difference = (
        destination_df["Recommended_Trip_Duration"] -
        duration
    ).abs()

    duration_score = (
        1 -
        duration_difference /
        destination_df["Recommended_Trip_Duration"].max()
    ).clip(0, 1)


    # ---------------------------------------------------------
    # 3. Calculate season match
    # ---------------------------------------------------------
    # 1 = matching season
    # 0 = different season

    season_score = (
        destination_df["Best_Season"] == season
    ).astype(int)


    # ---------------------------------------------------------
    # 4. Calculate category match
    # ---------------------------------------------------------
    # 1 = matching category
    # 0 = different category

    category_score = (
        destination_df["Category"] == category
    ).astype(int)


    # ---------------------------------------------------------
    # 5. Normalize Adventure Score
    # ---------------------------------------------------------
    # Convert the original 2-9 scale into 0-1
    # so it can be combined with the other scores.

    adventure_score = (
        destination_df["Adventure_Score"] -
        destination_df["Adventure_Score"].min()
    ) / (
        destination_df["Adventure_Score"].max() -
        destination_df["Adventure_Score"].min()
    )


    # ---------------------------------------------------------
    # 6. Define recommendation weights
    # ---------------------------------------------------------
    # These are our baseline weights.
    #
    # Season and category are given the highest importance
    # because they represent strong user preferences.

    season_weight = 0.30
    category_weight = 0.30
    budget_weight = 0.20
    duration_weight = 0.10
    adventure_weight = 0.10


    # ---------------------------------------------------------
    # 7. Calculate final recommendation score
    # ---------------------------------------------------------

    recommendation_score = (
        season_score * season_weight
        + category_score * category_weight
        + budget_score * budget_weight
        + duration_score * duration_weight
        + adventure_score * adventure_weight
    )


    # ---------------------------------------------------------
    # 8. Create the result DataFrame
    # ---------------------------------------------------------

    results = destination_df[
        [
            "Destination",
            "State",
            "Category",
            "Best_Season",
            "Average_Budget",
            "Recommended_Trip_Duration",
            "Adventure_Score"
        ]
    ].copy()


    # Add the calculated recommendation score.

    results["Recommendation_Score"] = recommendation_score


    # ---------------------------------------------------------
    # 9. Sort destinations by recommendation score
    # ---------------------------------------------------------

    results = results.sort_values(
        by="Recommendation_Score",
        ascending=False
    )


    # ---------------------------------------------------------
    # 10. Return only the requested number of results
    # ---------------------------------------------------------

    return results.head(top_n).reset_index(drop=True)
```

```python
# Create multiple test cases to evaluate our recommendation system.
#
# Each test case represents a different type of traveler.

test_cases = [
    {
        "name": "Adventure Traveler",
        "budget": 35000,
        "duration": 5,
        "season": "Mar-Jun",
        "category": "Adventure"
    },
    {
        "name": "Beach Traveler",
        "budget": 30000,
        "duration": 4,
        "season": "Nov-Feb",
        "category": "Beach"
    },
    {
        "name": "Heritage Traveler",
        "budget": 25000,
        "duration": 3,
        "season": "Oct-Mar",
        "category": "Heritage"
    },
    {
        "name": "Spiritual Traveler",
        "budget": 25000,
        "duration": 3,
        "season": "Oct-Mar",
        "category": "Spiritual"
    }
]

# Display the test cases so we can verify them.
pd.DataFrame(test_cases)
```

```python
# Run the recommendation system for every test case.
#
# We will store the best recommendation and its score
# so that we can compare the results easily.

evaluation_results = []

for test in test_cases:

    # Generate recommendations using the current test case.
    recommendations = recommend_destinations(
        budget=test["budget"],
        duration=test["duration"],
        season=test["season"],
        category=test["category"],
        top_n=5
    )

    # Get the highest-ranked recommendation.
    best = recommendations.iloc[0]

    # Store the evaluation information.
    evaluation_results.append({
        "Test Case": test["name"],
        "Requested Category": test["category"],
        "Requested Season": test["season"],
        "Best Recommendation": best["Destination"],
        "Recommended Category": best["Category"],
        "Recommended Season": best["Best_Season"],
        "Recommendation Score": best["Recommendation_Score"]
    })


# Convert the evaluation results into a DataFrame.
evaluation_df = pd.DataFrame(evaluation_results)

# Display the evaluation results.
evaluation_df
```
