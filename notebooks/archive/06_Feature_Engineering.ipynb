### Feature Engineering On combined destination weather and holidays dataset

```python
# ---------------------------------------------------------
# FEATURE ENGINEERING
# ---------------------------------------------------------

# Import pandas for data manipulation and analysis.
import pandas as pd

# Import numpy for numerical operations.
import numpy as np


# Define the path of the integrated destination dataset.
# This dataset contains our original destination features
# together with the weather features collected from OpenWeather.

destination_weather_path = (
    "../data/cleaned/destination_weather_features.csv"
)


# Load the integrated dataset into a DataFrame.

destination_df = pd.read_csv(
    destination_weather_path
)


# Display the shape of the dataset.
# We expect 92 destinations and 55 columns.

print("Dataset shape:", destination_df.shape)


# Display the first five rows to verify that
# the dataset was loaded correctly.

destination_df.head()
```

```python
# Display all column names in the integrated dataset.
# This helps us understand which features are already available
# and prevents us from creating duplicate features.

print("Number of columns:", len(destination_df.columns))

print("\nColumns:")
for column in destination_df.columns:
    print(column)
```

```python
# Display the data type of every column.
#
# This helps us identify:
# - numerical columns
# - categorical columns
# - text columns
# - columns that may need conversion

print(destination_df.dtypes)
```

```python
# Count missing values in every column.

missing_values = destination_df.isnull().sum()


# Display only columns that contain missing values.
# Columns with zero missing values are not displayed.

print("Columns containing missing values:")
print(missing_values[missing_values > 0])
```

```python
# Identify numerical columns.

numerical_columns = destination_df.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()


# Identify categorical/text columns.

categorical_columns = destination_df.select_dtypes(
    include=["object", "bool"]
).columns.tolist()


# Display the results.

print("Number of numerical columns:",
      len(numerical_columns))

print("\nNumerical columns:")
print(numerical_columns)


print("\nNumber of categorical columns:",
      len(categorical_columns))

print("\nCategorical columns:")
print(categorical_columns)
```

```python
# Identify categorical/string columns in the dataset.
#
# We explicitly use "str" because our DataFrame contains
# pandas string columns. This avoids the warning generated
# by using "object" with newer pandas versions.

categorical_columns = destination_df.select_dtypes(
    include=["str"]
).columns.tolist()


# Display the number of categorical columns.

print("Number of categorical columns:",
      len(categorical_columns))


# Display the categorical columns.

print("\nCategorical columns:")
print(categorical_columns)
```

```python
# Create a binary feature indicating whether the current weather
# condition involves rain.
#
# 1 = Rain is present
# 0 = Rain is not present
#
# This can help the recommendation system identify destinations
# where outdoor activities may currently be less suitable.

destination_df["Rain_Indicator"] = (
    destination_df["Weather"]
    .str.lower()
    .eq("rain")
    .astype(int)
)


# Display a few rows to verify the new feature.

destination_df[
    ["Destination", "Weather", "Rain_Indicator"]
].head(10)
```

```python
# Define a comfortable temperature range for general travel.
#
# This is a project-level assumption:
# temperatures between 18°C and 30°C are considered
# comfortable for many general outdoor travel activities.
#
# 1 = Comfortable temperature
# 0 = Outside the comfortable range

destination_df["Comfortable_Temperature"] = (
    destination_df["Temperature"].between(18, 30)
    .astype(int)
)


# Display the result for a few destinations.

destination_df[
    ["Destination", "Temperature", "Comfortable_Temperature"]
].head(10)
```

```python
# Create a binary feature for high humidity.
#
# We consider humidity above 80% as high humidity.
#
# 1 = High humidity
# 0 = Normal humidity

destination_df["High_Humidity"] = (
    destination_df["Humidity"] > 80
).astype(int)


# Display the result for verification.

destination_df[
    ["Destination", "Humidity", "High_Humidity"]
].head(10)
```

```python
# Create a binary feature indicating heavily cloudy conditions.
#
# Cloudiness above 80% is considered heavily cloudy.
#
# 1 = Heavy cloud cover
# 0 = Lower cloud cover

destination_df["High_Cloudiness"] = (
    destination_df["Cloudiness"] > 80
).astype(int)


# Display the result for verification.

destination_df[
    ["Destination", "Cloudiness", "High_Cloudiness"]
].head(10)
```

```python
# ---------------------------------------------------------
# BUDGET FEATURE ENGINEERING
# ---------------------------------------------------------

# Display basic statistics for the Average_Budget column.
#
# This helps us understand the range and distribution of
# destination budgets before creating new budget features.

print("Budget statistics:")
print(destination_df["Average_Budget"].describe())
```

```python
# Display the minimum and maximum destination budgets.

print(
    "Minimum budget:",
    destination_df["Average_Budget"].min()
)

print(
    "Maximum budget:",
    destination_df["Average_Budget"].max()
)


# Display the median budget.
# The median is useful because it is less affected by
# extremely expensive destinations.

print(
    "Median budget:",
    destination_df["Average_Budget"].median()
)
```

```python
# Create a normalized budget score using Min-Max normalization.
#
# The cheapest destination will have a score close to 0.
# The most expensive destination will have a score close to 1.
#
# Formula:
#
# normalized_value =
#     (value - minimum) / (maximum - minimum)
#
# This allows destinations with different budget values
# to be compared on the same scale.

budget_min = destination_df["Average_Budget"].min()
budget_max = destination_df["Average_Budget"].max()


destination_df["Budget_Score"] = (
    (destination_df["Average_Budget"] - budget_min)
    / (budget_max - budget_min)
)


# Display a few destinations with their original budget
# and the newly created normalized score.

destination_df[
    [
        "Destination",
        "Average_Budget",
        "Budget_Score"
    ]
].head(10)
```

```python
# Divide destinations into three budget groups based on
# the distribution of Average_Budget.
#
# Using quantiles ensures that the groups are based on
# our actual dataset rather than arbitrary fixed thresholds.

destination_df["Budget_Category"] = pd.qcut(
    destination_df["Average_Budget"],
    q=3,
    labels=["Low", "Medium", "High"],
    duplicates="drop"
)


# Display the number of destinations in each budget category.

print(
    destination_df["Budget_Category"]
    .value_counts()
    .sort_index()
)
```

```python
# Create an indicator showing whether hotel pricing information
# is available for each destination.
#
# 1 = Hotel price data exists
# 0 = Hotel price data is missing

destination_df["Hotel_Price_Available"] = (
    destination_df["Average_Hotel_Price"]
    .notna()
    .astype(int)
)


# Create an indicator showing whether flight cost information
# is available.
#
# 1 = Flight cost data exists
# 0 = Flight cost data is missing

destination_df["Flight_Cost_Available"] = (
    destination_df["Average_Flight_Cost"]
    .notna()
    .astype(int)
)


# Display the new indicators.

destination_df[
    [
        "Destination",
        "Hotel_Price_Available",
        "Flight_Cost_Available"
    ]
].head(10)
```

```python
# Fill missing hotel numerical values using the median
# of the destinations where hotel data is available.
#
# We do NOT use zero because zero would incorrectly imply
# that hotels or hotel prices actually cost nothing.

hotel_numeric_columns = [
    "Average_Hotel_Rating",
    "Average_Hotel_Price",
    "Minimum_Hotel_Price",
    "Maximum_Hotel_Price",
    "Number_of_Hotels"
]


for column in hotel_numeric_columns:

    # Calculate the median using only available values.

    median_value = destination_df[column].median()

    # Fill missing values with the calculated median.

    destination_df[column] = destination_df[column].fillna(
        median_value
    )


# Fill missing flight numerical values using their
# respective medians.

flight_numeric_columns = [
    "Average_Flight_Cost",
    "Minimum_Flight_Cost",
    "Maximum_Flight_Cost",
    "Number_of_Routes"
]


for column in flight_numeric_columns:

    # Calculate the median from available flight records.

    median_value = destination_df[column].median()

    # Fill missing values with the median.

    destination_df[column] = destination_df[column].fillna(
        median_value
    )
```

```python
# Replace missing hotel city names with an explicit label.
#
# "No_Hotel_Data" tells the model that hotel information
# was unavailable rather than pretending that a city exists.

destination_df["Hotel_City"] = (
    destination_df["Hotel_City"]
    .fillna("No_Hotel_Data")
)


# Verify that Hotel_City no longer contains missing values.

print(
    "Missing Hotel_City values:",
    destination_df["Hotel_City"].isnull().sum()
)
```
