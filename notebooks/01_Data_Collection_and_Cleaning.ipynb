```python
import warnings
warnings.filterwarnings("ignore")

import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns
import missingno as msno

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)
```

```python
tourism = pd.read_csv("../data/raw/tourism_data.csv")

hotels = pd.read_csv("../data/raw/hotels_data.csv")

flights = pd.read_csv("../data/raw/flights.csv")

flight_cost = pd.read_csv("../data/raw/flight_cost.csv")
```

```python
print("Tourism :", tourism.shape)
print("Hotels :", hotels.shape)
print("Flights :", flights.shape)
print("Flight Cost :", flight_cost.shape)
```

```python
def dataset_summary(df, name):

    print("=" * 70)
    print(f"{name.upper()} DATASET")
    print("=" * 70)

    print(f"\nShape: {df.shape}")

    print("\nColumns:")
    print(df.columns.tolist())

    print("\nData Types:")
    print(df.dtypes)

    print("\nMissing Values:")
    print(df.isnull().sum())

    print("\nDuplicate Rows:")
    print(df.duplicated().sum())


    print("\nFirst 5 Rows:")
    display(df.head())
```

```python
dataset_summary(tourism, "Tourism")
```

```python
dataset_summary(hotels, "Hotels")
```

```python
dataset_summary(flights, "Flights")
```

```python
dataset_summary(flight_cost, "Flight Cost")
```

```python
hotels[hotels.duplicated()]
```

```python
hotel_duplicates = hotels[hotels.duplicated(keep=False)]
hotel_duplicates.sort_values(by=hotels.columns.tolist())
```

```python
flight_duplicates = flights[flights.duplicated(keep=False)]
flight_duplicates.sort_values(by=flights.columns.tolist())
```

```python
# Hotels
hotels = hotels.drop_duplicates().reset_index(drop=True)

# Flights
flights = flights.drop_duplicates().reset_index(drop=True)
```

```python
print(hotels.shape)
print(flights.shape)
```

```python
tourism.to_csv("../data/cleaned/tourism_data_clean.csv", index=False)

hotels.to_csv("../data/cleaned/hotels_data_clean.csv", index=False)

flights.to_csv("../data/cleaned/flights_clean.csv", index=False)

flight_cost.to_csv("../data/cleaned/flight_cost_clean.csv", index=False)
```
